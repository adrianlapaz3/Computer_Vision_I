### Coordenadas cromáticas

```python
CoordCrom_1 = cv2.imread("coord_cromaticas/CoordCrom_1.png")
CoordCrom_2 = cv2.imread("coord_cromaticas/CoordCrom_2.png")
CoordCrom_3 = cv2.imread("coord_cromaticas/CoordCrom_3.png")
```

```python
# Pasar de BGR a RGB porque las imagen en Matplotlib se muestran en RGB
CoordCrom_1 = cv2.cvtColor(CoordCrom_1, cv2.COLOR_BGR2RGB)
CoordCrom_2 = cv2.cvtColor(CoordCrom_2, cv2.COLOR_BGR2RGB)
CoordCrom_3 = cv2.cvtColor(CoordCrom_3, cv2.COLOR_BGR2RGB)
```

```python
# Coordenadas cromaticas
def chromatic_coordinates(img):
    img = img.astype(np.float32)
    sum_channels = np.sum(img, axis=2, keepdims=True)
    sum_channels[sum_channels == 0] = 1e-6
    return img / sum_channels

chromatic_1 = chromatic_coordinates(CoordCrom_1)
chromatic_2 = chromatic_coordinates(CoordCrom_2)
chromatic_3 = chromatic_coordinates(CoordCrom_3)
```

```python
fig_CoordCrom, ax = plt.subplots(nrows=2, ncols=3, figsize=(15, 10))

ax[0, 0].imshow(CoordCrom_1)
ax[0, 0].set_title("Imagen original RGB 1")
ax[0, 0].axis("off")

ax[0, 1].imshow(CoordCrom_2)
ax[0, 1].set_title("Imagen original RGB 2")
ax[0, 1].axis("off")

ax[0, 2].imshow(CoordCrom_3)
ax[0, 2].set_title("Imagen original RGB 3")
ax[0, 2].axis("off")

ax[1, 0].imshow(chromatic_1)
ax[1, 0].set_title("Coordenadas cromáticas 1")
ax[1, 0].axis("off")

ax[1, 1].imshow(chromatic_2)
ax[1, 1].set_title("Coordenadas cromáticas 2")
ax[1, 1].axis("off")

ax[1, 2].imshow(chromatic_3)
ax[1, 2].set_title("Coordenadas cromáticas 3")
ax[1, 2].axis("off")

plt.show();
```

### White Patch

```python
# White Patch
def white_patch(img):
    img = img.astype(np.float32)
    max_vals = np.max(img, axis=(0, 1))
    scale = 255.0 / max_vals
    return np.clip(img * scale, 0, 255).astype(np.uint8)

# Pasar de RGB a White Patch
```

```python
white_patch_blue = white_patch(test_blue)
white_patch_green = white_patch(test_green) 
white_patch_red = white_patch(test_red)

white_patch_red2 = white_patch(wp_red_ok)
white_patch_green2 = white_patch(wb_green)  

white_patch_red3 = white_patch(wp_red2_ok)
white_patch_green3 = white_patch(wb_green2)
white_patch_blue3 = white_patch(wb_blue2)
```

Tras el recorte, se observa que ambas imágenes presentan una distribución de niveles digitales notablemente similar, lo que permite realizar una descripción general del histograma que es válida para ambas. ### Segmentar la imagen
Al observar los histogramas y los ejes cartesianos de la Figura original, se puede partir por lo más simple. En este caso, el suelo se diferencia claramente del cielo y agua porque todos sus niveles digitales son mayores en el canal rojo con respecto al verde y azul. Luego, el cielo también es fácil de diferenciar, todos los píxeles que presenten valores de rojo menores al los otros dos canales, y a su vez estén por debajo de 185 en la coordenada cartesiana y, son cielo. El resto, es todo agua.

```python
height, width = segmentacion.shape[:2]

# Coordenadas Y
y_coords = np.indices((height, width))[0]  

# Regla 1: R > B -> Suelo
mask_suelo = (segmentacion[:,:,0] > segmentacion[:,:,2])

# Regla 2: Si no es suelo y Y <= 285 -> Cielo
mask_cielo = (~mask_suelo) & (y_coords <= 285)

# Regla 3: El resto -> Agua
mask_agua = (~mask_suelo) & (y_coords > 285)

# Aplicar máscaras
suelo = segmentacion * mask_suelo[..., np.newaxis]
cielo = segmentacion * mask_cielo[..., np.newaxis]
agua = segmentacion * mask_agua[..., np.newaxis]


# Mostrar resultados
fig, ax = plt.subplots(2, 2, figsize=(10, 8))

ax[0,0].imshow(segmentacion)
ax[0,0].set_title('Original')
ax[0,0].axis('off')

ax[0,1].imshow(cielo)
ax[0,1].set_title('Cielo (Y ≤ 285)')
ax[0,1].axis('off')

ax[1,0].imshow(suelo)
ax[1,0].set_title('Suelo (R > B)')
ax[1,0].axis('off')

ax[1,1].imshow(agua)
ax[1,1].set_title('Agua (Resto)')
ax[1,1].axis('off')

plt.tight_layout()
plt.show();
```
